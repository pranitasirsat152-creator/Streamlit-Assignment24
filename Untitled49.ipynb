```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load Heart Dataset

heart = pd.read_csv("heart.csv")

print("Heart Dataset Shape:", heart.shape)
print(heart.head())

# Check Missing Values
print("\nMissing Values:")
print(heart.isnull().sum())

# Remove Duplicates
heart = heart.drop_duplicates()

# Encode Categorical Columns
le = LabelEncoder()

for col in heart.select_dtypes(include="object").columns:
    heart[col] = le.fit_transform(heart[col])

# Features & Target
X_cls = heart.drop("HeartDisease", axis=1)
y_cls = heart["HeartDisease"]

# Scaling
scaler_cls = StandardScaler()
X_cls = scaler_cls.fit_transform(X_cls)

# Train Test Split
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42
)

print("\nHeart Dataset Ready")


# Load Ford Dataset
ford = pd.read_csv("ford_car_dataset.csv")

print("\nFord Dataset Shape:", ford.shape)
print(ford.head())

# Check Missing Values
print("\nMissing Values:")
print(ford.isnull().sum())

# Remove Duplicates
ford = ford.drop_duplicates()

# Encode Categorical Columns
for col in ford.select_dtypes(include="object").columns:
    ford[col] = le.fit_transform(ford[col])

# Features & Target
X_reg = ford.drop("price", axis=1)
y_reg = ford["price"]

# Scaling
scaler_reg = StandardScaler()
X_reg = scaler_reg.fit_transform(X_reg)

# Train Test Split
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("\nFord Dataset Ready")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score
import pandas as pd

# ==========================
# Classification Models
# ==========================

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM": SVC(),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB()
}

results = []
trained_models = {}

for name, model in models.items():

    # Train Model
    model.fit(X_train_cls, y_train_cls)

    # Prediction
    y_pred = model.predict(X_test_cls)

    # Accuracy
    accuracy = accuracy_score(y_test_cls, y_pred)

    print("="*50)
    print("Model :", name)
    print("Accuracy :", accuracy)

    results.append({
        "Model": name,
        "Accuracy": accuracy
    })

    trained_models[name] = model


# ==========================
# Comparison Table
# ==========================

comparison = pd.DataFrame(results)

comparison = comparison.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

print("\nModel Comparison")
print(comparison)

# ==========================
# Best Model
# ==========================

best_model_name = comparison.iloc[0]["Model"]
best_accuracy = comparison.iloc[0]["Accuracy"]

best_classification_model = trained_models[best_model_name]

print("\nBest Classification Model :", best_model_name)
print("Best Accuracy :", best_accuracy)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import pandas as pd

# ==========================
# Regression Models
# ==========================

regression_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "Support Vector Regressor": SVR(),
    "KNN Regressor": KNeighborsRegressor()
}

regression_results = []
trained_regression_models = {}

for name, model in regression_models.items():

    # Train Model
    model.fit(X_train_reg, y_train_reg)

    # Prediction
    y_pred = model.predict(X_test_reg)

    # Evaluation
    r2 = r2_score(y_test_reg, y_pred)
    mae = mean_absolute_error(y_test_reg, y_pred)
    mse = mean_squared_error(y_test_reg, y_pred)

    print("="*60)
    print("Model :", name)
    print("R2 Score :", r2)
    print("MAE :", mae)
    print("MSE :", mse)

    regression_results.append({
        "Model": name,
        "R2 Score": r2,
        "MAE": mae,
        "MSE": mse
    })

    trained_regression_models[name] = model


# ==========================
# Comparison Table
# ==========================

comparison_reg = pd.DataFrame(regression_results)

comparison_reg = comparison_reg.sort_values(
    by="R2 Score",
    ascending=False
).reset_index(drop=True)

print("\nRegression Model Comparison")
print(comparison_reg)

# ==========================
# Best Regression Model
# ==========================

best_regression_name = comparison_reg.iloc[0]["Model"]
best_regression_score = comparison_reg.iloc[0]["R2 Score"]

best_regression_model = trained_regression_models[best_regression_name]

print("\nBest Regression Model :", best_regression_name)
print("Best R2 Score :", best_regression_score)
```

```python
import streamlit as st
import joblib
import pandas as pd

st.set_page_config(
    page_title="Heart Disease Prediction",
    layout="centered"
)

st.title("❤️ Heart Disease Prediction App")
st.write("AIML Assignment 24")

# ==========================
# Load Saved Model
# ==========================
model = joblib.load("best_classification_model.pkl")

# ==========================
# User Inputs
# ==========================
age = st.number_input("Age", 1, 100, 40)

sex = st.selectbox("Sex", ["M", "F"])

chest = st.selectbox(
    "Chest Pain Type",
    ["ATA", "NAP", "ASY", "TA"]
)

bp = st.number_input(
    "Resting Blood Pressure",
    50,
    250,
    120
)

chol = st.number_input(
    "Cholesterol",
    50,
    600,
    200
)

fasting = st.selectbox(
    "Fasting Blood Sugar",
    [0, 1]
)

ecg = st.selectbox(
    "Resting ECG",
    ["Normal", "ST", "LVH"]
)

hr = st.number_input(
    "Maximum Heart Rate",
    50,
    250,
    150
)

angina = st.selectbox(
    "Exercise Angina",
    ["N", "Y"]
)

oldpeak = st.number_input(
    "Old Peak",
    0.0,
    10.0,
    0.0
)

slope = st.selectbox(
    "ST Slope",
    ["Up", "Flat", "Down"]
)

# ==========================
# Manual Encoding
# ==========================
sex = 1 if sex == "M" else 0

chest_map = {
    "ATA": 0,
    "NAP": 1,
    "ASY": 2,
    "TA": 3
}

ecg_map = {
    "Normal": 0,
    "ST": 1,
    "LVH": 2
}

angina_map = {
    "N": 0,
    "Y": 1
}

slope_map = {
    "Up": 0,
    "Flat": 1,
    "Down": 2
}

input_data = pd.DataFrame([[
    age,
    sex,
    chest_map[chest],
    bp,
    chol,
    fasting,
    ecg_map[ecg],
    hr,
    angina_map[angina],
    oldpeak,
    slope_map[slope]
]], columns=[
    "Age",
    "Sex",
    "ChestPainType",
    "RestingBP",
    "Cholesterol",
    "FastingBS",
    "RestingECG",
    "MaxHR",
    "ExerciseAngina",
    "Oldpeak",
    "ST_Slope"
])

# ==========================
# Prediction
# ==========================
if st.button("Predict"):

    prediction = model.predict(input_data)

    if prediction[0] == 1:
        st.error("⚠️ Heart Disease Detected")
    else:
        st.success("✅ No Heart Disease")
```
